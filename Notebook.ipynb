```python
import pandas as pd
from IPython.display import display
df = pd.read_csv("animes.csv")

# 1. Conversion des notes en nombres pour permettre les calculs
for col in ['Note_Globale', 'Note_Meilleur_Ep', 'Note_Pire_Ep']:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# 2. Gestion des épisodes (on remplace les vides par 0)
df['Nb_Episodes'] = pd.to_numeric(df['Nb_Episodes'], errors='coerce').fillna(0)

# 3. Suppression des lignes sans "Note_Globale"
df = df.dropna(subset=['Note_Globale'])

df_nettoye = df.dropna(how='all')
df_nettoye.to_csv("animes_propre.csv", index=False)
print("\nFichier 'animes_propre.csv' créé avec succès !")
```

```python
df["Ecart"]=df["Note_Meilleur_Ep"]-df["Note_Pire_Ep"]
df["Regularite"]=10 - df["Ecart"]
df["Score_Metier"]=(df["Note_Globale"] * 0.7) + (df["Regularite"] * 0.3) 
moyenne_metier=df["Score_Metier"].mean()
print(f"La Moyenne c'est : {moyenne_metier:.2f}/10")
score_max=df["Score_Metier"].max()
meilleur_anime=df[df["Score_Metier"]==score_max]["Anime"].values[0]
print(f"l'anime numero 1 est: {meilleur_anime}")
display(df[["Anime","Note_Globale","Regularite","Score_Metier"]].head(10))
```

```python
import numpy as np
import plotly.express as px

# 1. Calcul de la Régularité et du Score Qualité
df['Regularite'] = 10 - (df['Note_Meilleur_Ep'] - df['Note_Pire_Ep'])
df['Score_Qualite'] = (df['Note_Globale'] * 0.6) + (df['Regularite'] * 0.4)

# 2. Application des coefficients selon la durée (Nb_Episodes)
def calculer_coefficient(nb_ep):
    if nb_ep <= 150: return 1.0
    elif 150 < nb_ep <= 300: return 0.85
    else: return 0.75

df['Coefficient'] = df['Nb_Episodes'].apply(calculer_coefficient)

# 3. Calcul du Score Final de Recommandation (%)
df['Recommandation_Pct'] = df['Score_Qualite'] * df['Coefficient'] * 10

# 4. Top 10 Interactif (Plotly)
top_10_rec = df.sort_values(by='Recommandation_Pct', ascending=False).head(10)

fig = px.bar(top_10_rec, 
             x='Recommandation_Pct', 
             y='Anime', 
             orientation='h',
             title="Top 10 des Animés les plus recommandés (Score Qualité Pondéré)",
             labels={'Recommandation_Pct': 'Pourcentage de Recommandation (%)'},
             color='Recommandation_Pct',
             color_continuous_scale='Viridis')

fig.show()

# 5. Export final conforme à ton README
df.to_csv("anime_nettoye.csv", index=False)
```
